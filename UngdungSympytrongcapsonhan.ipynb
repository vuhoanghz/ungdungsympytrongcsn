```python
# CHƯƠNG 2: CÁC LỆNH SYMPY CƠ BẢN

# Thay số cho nhiều biến cùng lúc
import sympy as sp
sp.init_printing()

u1, q, n = sp.symbols('u1 q n', real=True)
un_formula = u1 * q**(n - 1)

# Example: u1 = 3, q = 2, find u5
ket_qua = un_formula.subs({u1: 3, q: 2, n: 5})
print(ket_qua)

# Chuyển sang số thập phân
bieu_thuc = sp.sqrt(2) + sp.Rational(1, 3)
print(bieu_thuc)             # sqrt(2) + 1/3
print(bieu_thuc.evalf())     # 1.74754689571453
print(bieu_thuc.evalf(20))   # 1.7475468957145320588

# Các lệnh biến đổi biểu thức
q = sp.symbols('q', real=True)

expr = (q**3 - 1) / (q - 1)
print("simplify:", sp.simplify(expr))     # q**2 + q + 1
print("expand:  ", sp.expand((q + 1)**3)) # q**3 + 3*q**2 + 3*q + 1
print("factor:  ", sp.factor(q**2 - 1))   # (q - 1)*(q + 1)

# Tạo và quan sát phương trình
u1, q = sp.symbols('u1 q', real=True)

pt = sp.Eq(u1 + u1*q, 6)
print(pt)              # Eq(q*u1 + u1, 6)
print(pt.lhs)          # q*u1 + u1  (ve trai)
print(pt.rhs)          # 6          (ve phai)

# Giải một phương trình
u1, q = sp.symbols('u1 q', real=True)
pt = sp.Eq(u1 + u1*q, 6)
print(sp.solve(pt, u1))     # [6/(q + 1)]
```

```python
# CHƯƠNG 3: CÁC DẠNG TOÁN CSN CƠ BẢN

# Ví dụ 1: Tìm u5 khi u1 = 3, q = 2
import sympy as sp
sp.init_printing()

u1, q, n = sp.symbols('u1 q n', real=True)
un_formula = u1 * q**(n - 1)

# u1 = 3, q = 2, find u5
ket_qua = un_formula.subs({u1: 3, q: 2, n: 5})
print(ket_qua)

# Ví dụ 2: Tìm u10 khi u1 = 1, q = 3
ket_qua = un_formula.subs({u1: 1, q: 3, n: 10})
print(ket_qua)      # 19683

# Ví dụ 3: Tính S6 khi u1 = 3, q = 2
Sn_formula = u1 * (1 - q**n) / (1 - q)

S6 = Sn_formula.subs({u1: 3, q: 2, n: 6})
print(S6)

# Tính S10 khi u1 = 1, q = 1/2
u1, q, n = sp.symbols('u1 q n', real=True)
Sn_formula = u1 * (1 - q**n) / (1 - q)

S10 = Sn_formula.subs({u1: 1, q: sp.Rational(1, 2), n: 10})
print("Exact fraction:", S10)       # 1023/512
print("Decimal (10 digits):", S10.evalf(10))  # 1.998046875

# Hàm tính Sn có xử lý riêng trường hợp q = 1
def tinh_Sn(u1_val, q_val, n_val):
    """Compute S_n, handling the special case q = 1."""
    if q_val == 1:
        return n_val * u1_val
    else:
        u1s, qs, ns = sp.symbols('u1 q n')
        cong_thuc = u1s * (1 - qs**ns) / (1 - qs)
        return cong_thuc.subs({u1s: u1_val, qs: q_val, ns: n_val})

print("q = 1 case:", tinh_Sn(5, 1, 10))   # 5*10 = 50
print("q = 3 case:", tinh_Sn(5, 3, 4))    # 200
```

```python
# CHƯƠNG 4: GIẢI HỆ PHƯƠNG TRÌNH TÌM u1, q

# Thiết lập hệ phương trình từ đề bài
import sympy as sp
sp.init_printing()

u1, q = sp.symbols('u1 q', real=True)

# Build general terms for the required indices
u2 = u1 * q**(2 - 1)   # = u1*q
u3 = u1 * q**(3 - 1)   # = u1*q**2
u4 = u1 * q**(4 - 1)   # = u1*q**3
u5 = u1 * q**(5 - 1)   # = u1*q**4

# Given: u2 + u4 = 30  and  u3 + u5 = 90
pt1 = sp.Eq(u2 + u4, 30)
pt2 = sp.Eq(u3 + u5, 90)

print(pt1)
print(pt2)

# Giải hệ phương trình tìm u1, q
nghiem = sp.solve([pt1, pt2], [u1, q])
print(nghiem)

# Giải hệ bằng nonlinsolve
nghiem_nonlin = sp.nonlinsolve([pt1, pt2], [u1, q])
print(nghiem_nonlin)

# Giải phương trình bậc ba và lọc nghiệm thực
q = sp.symbols('q')

# Solve q**3 = 8
pt = sp.Eq(q**3, 8)
tat_ca_nghiem = sp.solve(pt, q)
print("All roots:", tat_ca_nghiem)

# Keep only real roots using .is_real
nghiem_thuc = [n for n in tat_ca_nghiem if n.is_real]
print("Real roots:", nghiem_thuc)

# Bài toán: tổng và tích ba số hạng liên tiếp
u1, q = sp.symbols('u1 q', real=True)

# Given: u1 + u2 + u3 = 14  and  u1 * u2 * u3 = 64
pt1 = sp.Eq(u1 + u1*q + u1*q**2, 14)
pt2 = sp.Eq(u1 * (u1*q) * (u1*q**2), 64)

tap_nghiem = sp.nonlinsolve([pt1, pt2], [u1, q])
print(tap_nghiem)

# Rút gọn biểu thức nghiệm
bt_phuc_tap = (u1**2 * q**2 - u1**2 * q**0) / (u1 * q - u1)
print("Before:", bt_phuc_tap)
print("After: ", sp.simplify(bt_phuc_tap))
```

```python
# ============================================================
# CHƯƠNG 5: CẤP SỐ NHÂN LÙI VÔ HẠN & GIỚI HẠN
# ============================================================

import sympy as sp
sp.init_printing()

u1, q, n = sp.symbols('u1 q n', real=True)

# Công thức tổng S_n
Sn_formula = u1 * (1 - q**n) / (1 - q)

# ------------------------------------------------------------
# CÁCH 1: Thay q bằng giá trị cụ thể (ví dụ q = 1/2) rồi mới lấy giới hạn
# ------------------------------------------------------------
Sn_vi_du = Sn_formula.subs(q, sp.Rational(1, 2))
S_vo_han_vi_du = sp.limit(Sn_vi_du, n, sp.oo)
print("With q = 1/2, S =", S_vo_han_vi_du)  # Kết quả: 2*u1

# ------------------------------------------------------------
# CÁCH 2: Hàm kiểm tra điều kiện và tính tổng trực tiếp
# ------------------------------------------------------------
def tong_CSN_lui_vo_han(u1_val, q_val):
    """Compute S = u1/(1-q), requires |q| < 1."""
    if abs(q_val) >= 1:
        raise ValueError(
            f"No finite sum since |q|={abs(q_val)} >= 1"
        )
    u1s, qs = sp.symbols('u1 q')
    return (u1s / (1 - qs)).subs({u1s: u1_val, qs: q_val})

# Ví dụ tính tổng chuỗi hình học hội tụ
print("S = 1 + 1/2 + 1/4 + ... =", tong_CSN_lui_vo_han(1, sp.Rational(1, 2)))   # Kết quả: 2
print("S = 3 - 1 + 1/3 - ... =", tong_CSN_lui_vo_han(3, sp.Rational(-1, 3)))  # Kết quả: 9/4
```

```python
# CHƯƠNG 6: ỨNG DỤNG THỰC TẾ

# Tính số dư sau n năm gửi tiết kiệm lãi kép
A, r, n = sp.symbols('A r n', positive=True)

# Balance after n periods at rate r per period
so_du = A * (1 + r)**n

# Deposit 100 million, 6%/year, after 10 years
so_du_thuc_te = so_du.subs({A: 100_000_000, r: sp.Rational(6, 100), n: 10})
print("Rounded (VND):", round(float(so_du_thuc_te.evalf()), 0))

# So sánh lãi đơn và lãi kép
A, r, n = sp.symbols('A r n', positive=True)

lai_don = A * (1 + n * r)         # simple interest
lai_kep = A * (1 + r)**n          # compound interest

# Compare after 20 years with A = 100, r = 5%/year
A_val, r_val, n_val = 100, sp.Rational(5, 100), 20
print("Simple:  ", lai_don.subs({A: A_val, r: r_val, n: n_val}))
print("Compound:", lai_kep.subs({A: A_val, r: r_val, n: n_val}).evalf(6))

# So sánh tăng trưởng vi khuẩn với hai tốc độ
u1_val = 100    # initial 100 bacteria
q1 = sp.Rational(3, 2)   # +50% per hour
q2 = 2                   # doubles per hour

u1s, qs, ns = sp.symbols('u1 q n')
cong_thuc = u1s * qs**(ns - 1)

for gio in range(1, 6):
    sl1 = cong_thuc.subs({u1s: u1_val, qs: q1, ns: gio})
    sl2 = cong_thuc.subs({u1s: u1_val, qs: q2, ns: gio})
    print(f"Hour {gio}: +50% -> {sl1},  doubled -> {sl2}")

# Dự báo dân số với tỉ lệ tăng 1.5%/năm
u1s, qs, ns = sp.symbols('u1 q n')
cong_thuc = u1s * qs**(ns - 1)

# Year 2025: 2 million, growth 1.5%/year, predict 2040
ket_qua = cong_thuc.subs({u1s: 2, qs: sp.Rational(15, 1000) + 1, ns: 15})
print("Population 2040 (million):", ket_qua.evalf(6))
```

```python
# CHƯƠNG 7: TRỰC QUAN HÓA VỚI MATPLOTLIB

# Vẽ đồ thị so sánh các CSN với q khác nhau
import matplotlib.pyplot as plt
import sympy as sp

u1s, qs, ns = sp.symbols('u1 q n')
cong_thuc = u1s * qs**(ns - 1)

# Convert symbolic expression to fast numeric function
f = sp.lambdify((u1s, qs, ns), cong_thuc, 'numpy')

so_hang = list(range(1, 11))     # n = 1..10
danh_sach_q = [0.5, 1.2, 1.5, 2] # multiple ratios

plt.figure(figsize=(8, 5))
for q_val in danh_sach_q:
    gia_tri = [f(10, q_val, n_val) for n_val in so_hang]
    plt.plot(so_hang, gia_tri, marker='o', label=f"q = {q_val}")

plt.xlabel("Term index n")
plt.ylabel("Value u_n (with u1 = 10)")
plt.title("Growth comparison of geometric progressions")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig("growth_chart_example.png", dpi=150)
plt.show()

# So sánh dãy rời rạc và hàm mũ liên tục
import numpy as np
import matplotlib.pyplot as plt

q = 2
u1 = 1

# Discrete points
n_int = np.arange(1, 9)
u_int = [u1 * q**(k - 1) for k in n_int]

# Continuous exponential curve
x_cont = np.linspace(1, 8, 200)
y_cont = u1 * q**(x_cont - 1)

plt.figure(figsize=(8, 5))
plt.plot(x_cont, y_cont, '--', color='gray',
         label=r'Continuous $y = 2^{x-1}$')
plt.scatter(n_int, u_int, color='red', s=60, zorder=3,
            label=r'Discrete terms $u_n$')
plt.xlabel('n')
plt.ylabel('u_n')
plt.title('Geometric progression as a discrete exponential function')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# CHƯƠNG 8: ĐÓNG GÓI MODULE & KIỂM THỬ

import sympy as sp


def tinh_un(u1_val, q_val, n_val):
    """Return u_n = u1 * q^(n-1)."""
    u1s, qs, ns = sp.symbols('u1 q n')
    return (u1s * qs**(ns - 1)).subs({u1s: u1_val, qs: q_val, ns: n_val})


def tinh_Sn(u1_val, q_val, n_val):
    """Return S_n; handles the special case q = 1."""
    if q_val == 1:
        return n_val * u1_val
    u1s, qs, ns = sp.symbols('u1 q n')
    ct = u1s * (1 - qs**ns) / (1 - qs)
    return ct.subs({u1s: u1_val, qs: q_val, ns: n_val})


def tong_lui_vo_han(u1_val, q_val):
    """Return S = u1/(1-q), requires |q| < 1."""
    if abs(q_val) >= 1:
        raise ValueError("Not convergent: need |q| < 1")
    u1s, qs = sp.symbols('u1 q')
    return (u1s / (1 - qs)).subs({u1s: u1_val, qs: q_val})


def giai_he_tim_u1_q(pt1, pt2, chi_lay_nghiem_thuc=True):
    """Solve 2 equations (sp.Eq) for (u1, q). Returns list of solutions."""
    u1, q = sp.symbols('u1 q', real=chi_lay_nghiem_thuc)
    nghiem = sp.solve([pt1, pt2], [u1, q])
    return nghiem


# test_solver.py -- kiểm thử đơn giản
# (Trong Colab, nếu đã chạy cell module ở trên, có thể bỏ dòng import này)
# from geometric_progression_solver import tinh_un, tinh_Sn, tong_lui_vo_han
import sympy as sp


def test_tinh_un():
    # u1 = 3, q = 2 -> u5 = 3 * 2^4 = 48
    assert tinh_un(3, 2, 5) == 48
    # u1 = 1, q = 1/2, n = 5 -> u5 = 1/16
    assert tinh_un(1, sp.Rational(1, 2), 5) == sp.Rational(1, 16)


def test_tinh_Sn_q_khac_1():
    # u1 = 3, q = 2 -> S6 = 189
    assert tinh_Sn(3, 2, 6) == 189


def test_tinh_Sn_q_bang_1():
    # q = 1 -> Sn = n * u1
    assert tinh_Sn(5, 1, 10) == 50


def test_tong_lui_vo_han():
    # u1 = 4, q = 1/2 -> S = 8
    assert tong_lui_vo_han(4, sp.Rational(1, 2)) == 8


if __name__ == "__main__":
    test_tinh_un()
    test_tinh_Sn_q_khac_1()
    test_tinh_Sn_q_bang_1()
    test_tong_lui_vo_han()
    print("All tests PASSED!")
```
